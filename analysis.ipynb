```python
import glob

all_files = glob.glob("*_youtube_trending_data.csv")
all_files
```

```python
import pandas as pd
import os

df_list = []

for file in all_files:
    country = file.split("_")[0]   # BR, CA, DE...
    temp = pd.read_csv(file)
    temp["country"] = country
    df_list.append(temp)

df = pd.concat(df_list, ignore_index=True)

df.shape
```

```python
df.isnull().sum()
```

```python
df['publishedAt'] = pd.to_datetime(df['publishedAt'], errors='coerce')
df['trending_date'] = pd.to_datetime(df['trending_date'], errors='coerce')
```

```python
df['publish_year'] = df['publishedAt'].dt.year
df['publish_month'] = df['publishedAt'].dt.month

df['trend_year'] = df['trending_date'].dt.year
df['trend_month'] = df['trending_date'].dt.month
```

```python
df['tags_list'] = df['tags'].apply(lambda x: str(x).split('|'))
```

```python
from collections import Counter

tag_counter = Counter()
df['tags_list'].apply(tag_counter.update)
tag_counter.most_common(20)
```

```python
import json

with open("US_category_id.json", "r") as f:
    data = json.load(f)

categories = {item['id']: item['snippet']['title'] for item in data['items']}
df['categoryId'] = df['categoryId'].astype(str)

df['category_name'] = df['categoryId'].map(categories)
```
